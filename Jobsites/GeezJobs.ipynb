```python
import requests
from bs4 import BeautifulSoup
```

```python
import csv
import time
```

```python
all_jobs = []
```

```python
for page_num in range(1, 12):
    site_map = "https://geezjobs.com/search-jobs?page="
    response = requests.get(site_map + str(page_num))
    soup = BeautifulSoup(response.content, 'html.parser')
    content = soup.find_all('div', class_ = 'row clients-page')
    for x in content:
      name = x.find('h4').text
      deadline = x.find('span',class_ = 'label label-red').text
      details =x.find('ul',class_='list-inline').text.replace('\n', '')#cleaned to remove new line
      link = x.find('a').get('href')
      fullinks = "https://geezjobs.com/" + link
      job_data = {
            "Job Title": name,
            "Deadline": deadline,
            "Details": details,
            "Link": fullinks
        }
      print(job_data)
      all_jobs.append(job_data)

```

```python
print(f" Total Jobs Scraped: {len(all_jobs)}")
```
