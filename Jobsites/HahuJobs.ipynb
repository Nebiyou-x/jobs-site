```python
import requests
from bs4 import BeautifulSoup
```

```python
job_data_list = []
```

```python
from urllib.parse import urljoin
```

```python
for page_num in range(1, 6):
    site_map = "https://www.hahu.jobs/jobs?page="
    response = requests.get(site_map + str(page_num))
    soup = BeautifulSoup(response.content, 'html.parser')
    content = soup.find_all('div', class_ = 'md:px-0')

    for x in content:
        h3_tag = x.find('h3')
        if not h3_tag:
            continue
        name = h3_tag.get_text(strip=True)

        det_element = x.find('span', {'itemprop': 'validThrough'})
        if not det_element:
            continue
        deadline = det_element.get_text(strip=True)

        detail_el = x.find('p', class_='text-base')
        details = detail_el.get_text(strip=True) if detail_el else "No details"

        link_tag = x.find('a', itemprop='url')  # not all div elem contain the details
        link = urljoin("https://www.hahu.jobs", link_tag['href']) if link_tag else "No link"


        job_data = {
            "Job Title": name,
            "Deadline": deadline,
            "Details": details,
            "Link": link
        }
        job_data_list.append(job_data)



for job in job_data_list:
    print(job)
```

```python
print(f"Scraped {len(job_data_list)} jobs")
```
